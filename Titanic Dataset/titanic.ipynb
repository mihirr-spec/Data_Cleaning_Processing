Why did we check the count?

Because a median calculated from a large group is generally more reliable than one calculated from a tiny group.

For example:

SibSp = 0

median = 29

This is based on a large amount of data.

But:

SibSp = 5
5 known Ages
median = 11

This is based on only 5 observations, so we should be cautious about using 11 as a representative age.

The most extreme example is:

SibSp = 8
0 known Ages
median = NaN

There are no known Ages in this group, so a median cannot be calculated.

Conclusion about SibSp

SibSp appears to contain some information about Age, but its groups are uneven in size.

Some groups have plenty of observations, while others have very few or none.

Therefore:

SibSp may be useful as an additional feature for estimating Age, but using SibSp alone to fill missing Age values could be unreliable.

## Investigating Missing Cabin Values by Pclass

We know that the `Cabin` column has a large number of missing values:

- Total rows: 891
- Missing Cabin values: 687
- Missing percentage: approximately 77.1%

Before deciding what to do with these missing values, we want to investigate whether the missingness is related to another variable.

We will investigate whether the percentage of missing `Cabin` values is different across the three passenger classes (`Pclass`).

### Command

df_clean.groupby("Pclass")["Cabin"].apply(lambda x: x.isnull().mean())

### What does this command do?

The entire DataFrame is first divided into groups based on `Pclass`.

Since `Pclass` has three possible values, we conceptually get three groups:

Entire DataFrame
↓
Group by Pclass
↓
Pclass = 1
Pclass = 2
Pclass = 3

We then select the `Cabin` column from each group.

Conceptually:

Pclass = 1 → Cabin values
Pclass = 2 → Cabin values
Pclass = 3 → Cabin values

The `["Cabin"]` part tells Pandas that we only want to work with the Cabin column.

### Understanding `apply()`

`.apply()` tells Pandas to run the given function separately on each Pclass group.

The function is:

lambda x: x.isnull().mean()

Here, `x` represents the `Cabin` Series belonging to the current Pclass group.

So conceptually:

Pclass = 1 → x = Cabin values for Pclass 1
Pclass = 2 → x = Cabin values for Pclass 2
Pclass = 3 → x = Cabin values for Pclass 3

### Understanding `isnull()`

The next operation is:

x.isnull()

This checks every Cabin value and returns a Boolean value.

If the Cabin value exists:

False

If the Cabin value is missing:

True

For example:

C85 → False
NaN → True
E46 → False
NaN → True
NaN → True

So `isnull()` converts the Cabin column into a Series of True and False values.

### Understanding `mean()`

We then use:

x.isnull().mean()

When calculating the mean of Boolean values, Pandas treats:

True = 1
False = 0

For example:

True
False
True
False
True

becomes:

1
0
1
0
1

The mean is therefore:

(1 + 0 + 1 + 0 + 1) / 5

= 3 / 5

= 0.60

Therefore:

0.60 = 60%

So the mean gives us the proportion of missing Cabin values.

### What are we calculating?

Because the operation is performed separately for each Pclass group, we get:

Pclass = 1 → percentage of missing Cabin values
Pclass = 2 → percentage of missing Cabin values
Pclass = 3 → percentage of missing Cabin values

This allows us to compare Cabin missingness across the three passenger classes.

### Overall Flow

Entire DataFrame
↓
Group by Pclass
↓
Three groups:
    Pclass = 1
    Pclass = 2
    Pclass = 3
↓
Select Cabin
↓
Apply function to each group
↓
isnull()
↓
True / False
↓
True = 1
False = 0
↓
mean()
↓
Proportion of missing Cabin values
↓
Compare missing Cabin percentage across Pclass

### One-Sentence Summary

We divide the dataset into three groups based on `Pclass`, select the `Cabin` column from each group, use `isnull()` to identify missing Cabin values as True/False, and calculate the mean of those Boolean values to determine the proportion of missing Cabin values within each passenger class.

### Why are we doing this?

We are trying to understand whether the missingness in `Cabin` is random or whether it is related to `Pclass`.

If, for example, first-class passengers have much less missing Cabin information than third-class passengers, then the missingness itself may contain useful information.

This is important because we should understand the missing data before deciding whether to:

- Fill the missing values
- Replace them with `"Unknown"`
- Extract another feature such as Deck
- Create a missing-value indicator
- Drop the original `Cabin` column
- Use another preprocessing strategy

We are currently in the investigation stage.

Detect → Investigate → Understand → Decide → Clean → Validate

## Handling Duplicate Deck Letters in Cabin Values

We previously used:

df_clean["Cabin"].str.findall(r"[A-G]")

This extracts every deck letter from each Cabin value.

For example:

C23 C25 C27 → ["C", "C", "C"]

F G73 → ["F", "G"]

F E69 → ["F", "E"]

A missing Cabin value → NaN

The problem is that when multiple cabins are on the same deck, we get duplicate letters.

For example:

C23 C25 C27 → ["C", "C", "C"]

We only need to record the deck once:

["C"]

However, we must preserve different decks.

For example:

F G73 → ["F", "G"]

F E69 → ["F", "E"]

Therefore, we need to remove duplicate deck letters while keeping different deck letters.

---

## 1. `dict.fromkeys(x)`

The syntax is:

dict.fromkeys(iterable)

It creates a dictionary using the values from `x` as dictionary keys.

Dictionary keys must be unique, so duplicate values are automatically removed.

For example:

x = ["C", "C", "C"]

dict.fromkeys(x)

produces:

{"C": None}

The duplicate `"C"` values disappear because dictionary keys must be unique.

Another example:

x = ["F", "G", "F"]

dict.fromkeys(x)

produces:

{"F": None, "G": None}

The second `"F"` is removed.

---

## 2. `list(...)`

We do not actually want a dictionary as our final result.

We want a list of deck letters.

So we use:

list(dict.fromkeys(x))

The process is:

["C", "C", "C"]
↓
dict.fromkeys(x)
↓
{"C": None}
↓
list(...)
↓
["C"]

Therefore:

dict.fromkeys(x) → removes duplicates

list(...) → converts the result back into a list

So:

["C", "C", "C"] → ["C"]

and:

["F", "G"] → ["F", "G"]

---

## 3. Understanding `isinstance()`

The syntax is:

isinstance(object, type)

It asks:

> Is this object an instance of this type?

For example:

x = ["C", "C", "C"]

isinstance(x, list)

returns:

True

because `x` is a list.

If:

x = "C85"

isinstance(x, list)

returns:

False

because `"C85"` is a string, not a list.

In our case:

isinstance(x, list)

means:

> Check whether `x` is a list.

---

## 4. Why Do We Need `isinstance()`?

Our `str.findall()` result contains two different types of values.

Passengers with Cabin information have lists:

["C"]

["C", "C", "C"]

["F", "G"]

Passengers with missing Cabin information have:

NaN

Therefore, we cannot blindly run:

list(dict.fromkeys(x))

on every value.

We only want to run it when `x` is actually a list.

So we use:

isinstance(x, list)

to check the type of `x`.

---

## 5. The `if / else` Expression

The syntax is:

A if CONDITION else B

This means:

> If the condition is True, use A. Otherwise, use B.

Our expression is:

list(dict.fromkeys(x)) if isinstance(x, list) else x

We can visualize it as:

isinstance(x, list)
        ↓
   ┌────┴────┐
   ↓         ↓
 True      False
   ↓         ↓
list(       return
dict...     x
)

If x is a list:

x = ["C", "C", "C"]

isinstance(x, list) → True

So Python executes:

list(dict.fromkeys(x))

Result:

["C"]

If x is not a list:

x = NaN

isinstance(x, list) → False

So Python skips the left side and simply returns:

x

Result:

NaN
